```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python


file_path1 = '/content/drive/MyDrive/Datasets/StudentsPerformance.csv'
file_path2 = '/content/drive/MyDrive/Datasets/gym_membership.csv'

students = pd.read_csv(file_path1)
gym = pd.read_csv(file_path2)

print(f'students: {students.shape}   gym: {gym.shape}')
```

```python
def eda(gym):

    fig, axes = plt.subplots(2, 3, figsize=(15, 8))

    ##1. Bar chart of membership types
    gym['abonoment_type'].value_counts().plot(
        kind='bar',
        ax=axes[0, 0],
        color='steelblue'
    )

    axes[0, 0].set_title('Membership Types')




    ##2. Histogram of avg_time_in_gym
    axes[0, 1].hist(
        gym['avg_time_in_gym'],
        bins=20,
        color='seagreen',
        edgecolor='white'
    )

    axes[0, 1].set_title('Average Time in Gym')





    ##3. Scatter plot of Age vs visit_per_week
    axes[0, 2].scatter(
        gym['Age'],
        gym['visit_per_week'],
        alpha=0.3
    )

    axes[0, 2].set(
        title='Age vs Visits Per Week',
        xlabel='Age',
        ylabel='Visits Per Week'
    )



    ##4. Box plot: gender vs Age
    sns.boxplot(
        x='gender',
        y='Age',
        data=gym,
        ax=axes[1, 0]
    )

    axes[1, 0].set_title('Gender vs Age')





    ##5. Count plot: days_per_week
    sns.countplot(
        x='days_per_week',
        data=gym,
        ax=axes[1, 1],
    )

    axes[1, 1].set_title('Workout Days Per Week')
    axes[1, 1].set_xticklabels([])





    ##6. Bar plot: mean visit_per_week per abonoment_type
    sns.barplot(
        x='abonoment_type',
        y='visit_per_week',
        data=gym,
        ax=axes[1, 2]
    )

    axes[1, 2].set_title('Average Visits by Membership')



    plt.tight_layout()
    plt.show()


eda(gym)
```
