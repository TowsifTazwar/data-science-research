```python
#1 Write a function to calculate the sum of numbers from 1 to N.

def summation(n):
  sum=0
  for i in range(1,n+1):
    sum += i

  return sum

result= summation(10)
print(f"Sum: {result}")
```

```python
#2 Write a function that returns the square of a number.

def square(n):
  result= n**2
  return result

n=int(input("Enter the number: "))
answer= square(n)
print(f"Square of {n} is: {answer}")
```

```python
#3 Write a function that checks whether a number is even/odd.

def EvenOdd(n):
  if n % 2 ==0:
    result= "Even"
  else:
    result= "Odd"
  return result

n=int(input("Enter the number: "))
answer= EvenOdd(n)
print(f"The number is: {answer}")
```

```python
#4 Write a function that returns the larger number from two given numbers.

def Large(x,y):
  if x > y:
    result= x
  else:
    result= y
  return result

x=int(input("Enter First number: "))
y=int(input("Enter Second number: "))
answer= Large(x,y)
print(f"Largest Number: {answer}")
```

```python

#5  	Write a Python program to create a list of 10 integers. Then perform the following operations:
  #Append an element to the list.
  #Insert an element at a specific position.
  #Remove an element from the list.
  #Pop an element from a specific position.

list1=[]

print("Enter 10 integers: ")
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    list1.append(num)

print("Initial List:", list1)



```

```python
#Append an element to the list.

append_value= int(input("Enter an element to append: "))
list1.append(append_value)
print("List after appending:", list1)
```

```python
#Insert an element at a specific position.

insert_value= int(input("Enter an element to insert: "))
insert_position= int(input("Enter the position to insert: "))
list1.insert(insert_position, insert_value)
print("List after inserting:", list1)
```

```python
#Remove an element from the list.

remove_value= int(input("Enter an element to remove: "))
list1.remove(remove_value)
print("List after removing:", list1)
```

```python
#Pop an element from a specific position.

pop_position= int(input("Enter the position to pop: "))
popped_value= list1.pop(pop_position)
print("Popped value:", popped_value)
print("List after popping:", list1)
```

```python
#6 Write a program to find all unique elements in the list.

list2=[]

print("Enter 10 integers: ")
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    list2.append(num)

print("Initial List:", list2)

unique_elements = []
for element in list2:
    if element not in unique_elements:
        unique_elements.append(element)

print("Unique Elements:", unique_elements)

```

```python
#7 Write a Python program to calculate the mean of a list of numbers.

list3=[]

print("Enter 10 integers: ")
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    list3.append(num)

print("Initial List:", list3)

sum_of_numbers = __builtins__.sum(list3)
count_of_numbers = len(list3)

mean = sum_of_numbers / count_of_numbers

print("Mean:", mean)
```

```python
#8 Classify Numbers into Even and Odd Lists

list4=[]

print("Enter 10 integers: ")
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    list4.append(num)

print("Initial List:", list4)

list_of_even_numbers = []
list_of_odd_numbers = []

for num in list4:
  if num % 2 == 0:
    list_of_even_numbers.append(num)
  else:
    list_of_odd_numbers.append(num)

print("Even Numbers:", list_of_even_numbers)
print("Odd Numbers:", list_of_odd_numbers)

```

```python
## 9: Write a program to find the key in a dictionary that has the
# maximum value.
# Input: {‘a’: 10, ‘b’: 20, ‘c’: 15}
# Output: ‘b’


data = {'a': 10, 'b': 20, 'c': 15}


max_key = __builtins__.max(data, key=data.get)

print(max_key)
```

```python
#10: Write a program to create a dictionary where the keys are
#numbers from 1 to n and the values are their squares.
#Input: n = 5
#Output:
#{ 1: 1,
#2: 4,
#3: 9,
#4: 16,
#5: 25
#}


n = int(input("Enter a number: "))

squares = {}

for i in range(1, n + 1):
    squares[i] = i * i

print(squares)


```
